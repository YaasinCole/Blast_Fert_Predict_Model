```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.impute import SimpleImputer
import joblib

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.layers import LayerNormalization

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LayerNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Dense, LayerNormalization, Input
```

```python
## DATA PREPROCESSING
df = pd.read_csv('RBL Data Edited-Combined.csv', sep=";" , skiprows=1 , decimal=",", encoding="utf-8" , usecols=lambda col: col != 'Patient_ID')
## Dropping null values from the dataset
df = df.dropna(how="all")
df.isnull().sum()

#after the above we then separate the targets from input features also dropping unimportant features from the set
x = df.drop(columns=["Fertilization Rate (%)" , "Blastulation_Rate (%)","Year","Ejaculatory_Abstinence (days)"])
y = df[["Fertilization Rate (%)","Blastulation_Rate (%)"]]

## Added step for dealing with missing values 
num_cols= x.select_dtypes(include=["float64", "int64"]).columns
cat_cols= x.select_dtypes(include=["object"]).columns

num_imputer= SimpleImputer(strategy="median")
cat_imputer= SimpleImputer(strategy="most_frequent")

x[num_cols]= num_imputer.fit_transform(x[num_cols])
x[cat_cols]= cat_imputer.fit_transform(x[cat_cols])

## In the below we encode our categorical columns 
le_col= {}
for col in cat_cols:
        le = LabelEncoder()
        x[col] = le.fit_transform(x[col].astype(str))
        le_col[col] = le

## Convert encoded columns to dataframes to inspect 
# Making sure all the data is correct and no nulls for input features 
encoded_df = x.copy()
encoded_df.head()
print(encoded_df.dtypes)
print(encoded_df.describe())
print(encoded_df.isnull().sum())

## Making sure target variables are correct with no nulls 
print(y.head())          # sample values
print(y.dtypes)          # ensure nsumeric
print(y.nunique())
print(y.isnull().sum())

print(df.shape)
print("Total features:", len(x.columns))
```

```python
# Scale targets to (0,1) for beta regression
y_fert_binned = y["Fertilization Rate (%)"].str.replace("%","", regex=False).astype(float) / 100.0 
y_blast_binned = y["Blastulation_Rate (%)"].str.replace("%","", regex=False).astype(float) / 100.0

##this is when fertalisation rate is 0% and blastulation was left blank
structured_zero_mask = (y_fert_binned == 0) & (y_blast_binned.isna())

## Assigning blastulation to 0 
y_blast_binned.loc[structured_zero_mask] = 0.0

print("Structured zeros have been fixed", structured_zero_mask.sum())
```

```python
# =========================
# DATA SPLITTING
# =========================

# Use FULL dataset (no filtering)
X = encoded_df
original_feature_names = X.columns.tolist()
y_fert = y_fert_binned
y_blast = y_blast_binned

# 1. FIRST SPLIT: Temp + Holdout (20% holdout)
x_temp, x_holdout, y_fert_temp, y_fert_holdout, y_blast_temp, y_blast_holdout = train_test_split(
    X,
    y_fert,
    y_blast,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

# 2. SECOND SPLIT: Train + Test (30% of temp → test)
x_train, x_test, y_fert_train, y_fert_test, y_blast_train, y_blast_test = train_test_split(
    x_temp,
    y_fert_temp,
    y_blast_temp,
    test_size=0.3,
    random_state=42,
    shuffle=True
)

# 3. THIRD SPLIT: Train + Validation (20% of train → val)
x_train, x_val, y_fert_train, y_fert_val, y_blast_train, y_blast_val = train_test_split(
    x_train,
    y_fert_train,
    y_blast_train,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

# =========================
# SANITY CHECKS
# =========================
print("Data shapes after splitting:")
print("x_train shape:", x_train.shape)
print("x_val shape:", x_val.shape)
print("x_test shape:", x_test.shape)
print("x_holdout shape:", x_holdout.shape)

print("Fertilization (train):",
      y_fert_train.min(), y_fert_train.max(), np.mean(y_fert_train))
print("Blastulation (train):",
      y_blast_train.min(), y_blast_train.max(), np.mean(y_blast_train))
```

```python
#Zero fixed for beta loss
y_blast_train = y_blast_train.mask(y_fert_train ==0, 0.0)
y_blast_val = y_blast_val.mask(y_fert_val ==0, 0.0)

##Converting Panda series to numpy arrays
y_fert_train =np.array(y_fert_train , dtype=np.float32)
y_fert_val   =np.array(y_fert_val , dtype=np.float32)
y_blast_train = np.array(y_blast_train , dtype=np.float32)
y_blast_val   = np.array(y_blast_val , dtype=np.float32)

## Scaling my y-values 
epsilon = 1e-6
y_fert_train = np.clip(y_fert_train, epsilon, 1 - epsilon)
y_fert_val = np.clip(y_fert_val, epsilon, 1 - epsilon)

y_blast_train = np.clip(y_blast_train, epsilon, 1 - epsilon)
y_blast_val = np.clip(y_blast_val, epsilon, 1 - epsilon)

print("Scaled Fertilization range:", y_fert_train.min(), "→", y_fert_train.max())
print("Scaled Blastulation range:", y_blast_train.min(), "→", y_blast_train.max())
```

```python
##SCALING MY DATA 
## we will be capping the outliers within the 1st and the 99th percentiles 
def fit_outlier_caps(x, lower=0.01, upper=0.99):
    bounds = []
    x_np = x.to_numpy()
    for i in range(x_np.shape[1]):
        lo, hi = np.quantile(x_np[:, i], [lower, upper])
        bounds.append((lo, hi))
    return bounds

def apply_outlier_caps(x, bounds):
    x_np = x.to_numpy()
    for i, (lo, hi) in enumerate(bounds):
        x_np[:, i] = np.clip(x_np[:, i], lo, hi)
    return x_np


outlier_bounds = fit_outlier_caps(x_train)

x_train_capped = apply_outlier_caps(x_train, outlier_bounds)
x_val_capped = apply_outlier_caps(x_val, outlier_bounds)
x_test_capped = apply_outlier_caps(x_test, outlier_bounds)
x_holdout_capped = apply_outlier_caps(x_holdout, outlier_bounds)
## Initialising standard Scaler and scaling our input features for model training
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train_capped)
x_test_scaled = scaler.transform(x_test_capped)
x_val_scaled = scaler.transform(x_val_capped)
x_holdout_scaled = scaler.transform(x_holdout_capped)

## Visualising data before and after Scaling 
x_train_scaled_df = pd.DataFrame(x_train_scaled, columns=x_train.columns)
print("Before Scaling:\n", x_train.head())
print("\nAfter Scaling:\n", x_train_scaled_df.head())

assert x_train_scaled.shape[0] == y_fert_train.shape[0] == y_blast_train.shape[0]
assert x_val_scaled.shape[0] == y_fert_val.shape[0] == y_blast_val.shape[0]

print("All data shapes match!")
print('x_train_scaled shape: ', x_train_scaled.shape)
print('y_train_scaled shape: ', y_fert_train.shape)
print('y_train_scaled shape: ', y_blast_train.shape)
print(y_fert_train.shape, y_fert_train[:10], y_fert_train.dtype)
print(y_blast_train.shape, y_blast_train[:10], y_blast_train.dtype)

print("Final Fert range:", y_fert_train.min(), y_fert_train.max())
print("Final Blast range:", y_blast_train.min(), y_blast_train.max())
```

```python
##Baseline for 10 runs of the model
def build_fert_model(input_dim):
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(64, activation="relu"),
        LayerNormalization(),
        Dense(32, activation="relu"),
        Dense(1, activation="sigmoid")  # output scaled 0-1
    ])
    model.compile(optimizer=Adam(learning_rate=0.001), loss="mse")
    return model

def build_blast_model(input_dim):
    model = Sequential([
        Input(shape=(input_dim,)),
        Dense(64, activation="relu"),
        LayerNormalization(),
        Dense(32, activation="relu"),
        Dense(1, activation="sigmoid")
    ])
    model.compile(optimizer=Adam(learning_rate=0.001), loss="mse")
    return model
    

def run_baseline_n_times(x_train, y_train, x_val, y_val, n_runs=10):
    fert_mae_list = []
    blast_mae_list = []

    input_dim = x_train.shape[1]

    for i in range(n_runs):
        print(f"\n=== Run {i+1}/{n_runs} ===")
        
        # Fertilization model
        fert_model = build_fert_model(input_dim)
        fert_model.fit(
            x_train, y_train,
            validation_data=(x_val, y_val),
            epochs=100,
            batch_size=16,
            verbose=0,
            callbacks=[EarlyStopping(patience=10, restore_best_weights=True)]
        )
        y_pred = fert_model.predict(x_val, verbose=0)
        mae = mean_absolute_error(y_val, y_pred)
        fert_mae_list.append(mae)
        print(f"Fert MAE: {mae:.4f}")

    fert_mean = np.mean(fert_mae_list)
    fert_std = np.std(fert_mae_list)

    print("\n=== Baseline Fertilization Model ===")
    print(f"MAE over {n_runs} runs: mean = {fert_mean:.4f}, std = {fert_std:.4f}")

    return fert_mae_list, fert_mean, fert_std,fert_model


baseline_fert_mae_list, baseline_fert_mean, baseline_fert_std,fert_model = run_baseline_n_times(
    x_train_scaled, y_fert_train, x_val_scaled, y_fert_val, n_runs=10
)


# ===============================
# Feature-drop experiments
# ===============================
feature_importance_results = {}

for i, feature in enumerate(x_train.columns):
    print(f"\n--- Dropping feature: {feature} ---")
    
    # Drop column from train/val
    x_train_drop = np.delete(x_train_scaled, i, axis=1)
    x_val_drop   = np.delete(x_val_scaled, i, axis=1)
    
    # Run 10 times
    mae_list, mean_mae, std_mae,_ = run_baseline_n_times(x_train_drop, y_fert_train, x_val_drop, y_fert_val, n_runs=10)
    
    feature_importance_results[feature] = {
        "mean_mae": mean_mae,
        "std_mae": std_mae
    }

# Convert results to DataFrame for inspection
import pandas as pd
feat_imp_df = pd.DataFrame.from_dict(feature_importance_results, orient="index")
feat_imp_df.sort_values("mean_mae", ascending=False, inplace=True)
print("\nFeature drop results (higher MAE = more important):")
print(feat_imp_df)
```

```python
# ===============================
# Baseline + Feature-drop for Blastulation Model
# ===============================

def run_blast_baseline_n_times(x_train, y_train, x_val, y_val, n_runs=10):
    """
    Run blast model n times and return MAE metrics
    """
    blast_mae_list = []
    input_dim = x_train.shape[1]

    for i in range(n_runs):
        print(f"\n=== Run {i+1}/{n_runs} ===")
        
        # Build blast model
        blast_model = build_blast_model(input_dim)
        blast_model.fit(
            x_train, y_train,
            validation_data=(x_val, y_val),
            epochs=100,
            batch_size=16,
            verbose=0,
            callbacks=[EarlyStopping(patience=10, restore_best_weights=True)]
        )
        y_pred = blast_model.predict(x_val, verbose=0)
        mae = mean_absolute_error(y_val, y_pred)
        blast_mae_list.append(mae)
        print(f"Blast MAE: {mae:.4f}")

    mean_mae = np.mean(blast_mae_list)
    std_mae = np.std(blast_mae_list)

    print("\n=== Baseline Blastulation Model ===")
    print(f"MAE over {n_runs} runs: mean = {mean_mae:.4f}, std = {std_mae:.4f}")

    return blast_mae_list, mean_mae, std_mae

# --------------------------
# Prepare data with fert prediction as extra input
# --------------------------
fert_pred_train = fert_model.predict(x_train_scaled)[:, 0].reshape(-1, 1)
fert_pred_val = fert_model.predict(x_val_scaled)[:, 0].reshape(-1, 1)

x_train_blast = np.concatenate([x_train_scaled, fert_pred_train], axis=1)
x_val_blast = np.concatenate([x_val_scaled, fert_pred_val], axis=1)

blast_feature_names = original_feature_names + ["Predicted_Fertilization"]

# --------------------------
# Run baseline
# --------------------------
baseline_blast_mae_list, baseline_blast_mean, baseline_blast_std = run_blast_baseline_n_times(
    x_train_blast, y_blast_train, x_val_blast, y_blast_val, n_runs=10
)

# ===============================
# Feature-drop experiments
# ===============================
blast_feature_importance_results = {}

for i, feature_name in enumerate(blast_feature_names):
    print(f"\n--- Dropping feature:{feature_name} ---")
    
    x_train_drop = np.delete(x_train_blast, i, axis=1)
    x_val_drop = np.delete(x_val_blast, i, axis=1)
    
    mae_list, mean_mae, std_mae = run_blast_baseline_n_times(
        x_train_drop, y_blast_train, x_val_drop, y_blast_val, n_runs=10
    )
    
    blast_feature_importance_results[feature_name] = {
        "mean_mae": mean_mae,
        "std_mae": std_mae
    }

# Convert to DataFrame
blast_feat_imp_df = pd.DataFrame.from_dict(blast_feature_importance_results, orient="index")
blast_feat_imp_df.sort_values("mean_mae", ascending=False, inplace=True)
print("\nBlast Feature drop results (higher MAE = more important):")
print(blast_feat_imp_df)
```

```python
##Implementing beta negative likelihood (beta regression)
def beta_nll_loss(y_true, y_pred):
    mu = tf.clip_by_value(y_pred[:, 0], 1e-3, 1 - 1e-3)
    phi = tf.nn.softplus(y_pred[:, 1]) + 1e-3
    
    alpha = mu * phi
    beta = (1 - mu) * phi
    
    # Use log-beta function for stability
    log_beta = tf.math.lgamma(alpha) + tf.math.lgamma(beta) - tf.math.lgamma(alpha + beta)
    
    # Log likelihood using safe operations
    log_y = tf.math.log(tf.maximum(y_true, 1e-6))
    log_1_minus_y = tf.math.log(tf.maximum(1 - y_true, 1e-6))
    
    log_likelihood = (alpha - 1) * log_y + (beta - 1) * log_1_minus_y
    
    nll = log_beta - log_likelihood
    return tf.reduce_mean(nll)
```

```python
def evaluate_percent(y_true_scaled , p_pred_mu, name="model"):
# the y_true_scaled is in the 0-1 range we used in training as our target 
# y_pred_mu : the model predictions of(0-1 range)

    y_true_pct = y_true_scaled *100.0
    y_pred_pct = p_pred_mu  *100.0
    mae = mean_absolute_error(y_true_pct, y_pred_pct)
    rmse = math.sqrt(mean_squared_error(y_true_pct, y_pred_pct))
    print(f"{name}  | MAE (pct): {mae:.3f}  | RMSE (pct): {rmse:.3f}")
    return mae, rmse
```

```python
#BUILD NEURAL NETWORK 
#Clear any previous sessions 
tf.keras.backend.clear_session()

# Define OutputScaler layer
class OutputScaler(layers.Layer):
    def __init__(self, min_val, max_val, **kwargs):
        super().__init__(**kwargs)
        self.min_val = min_val
        self.max_val = max_val
        
    def call(self, inputs):
        mu = inputs[:, 0:1]
        # Scale mu to match target distribution
        mu_scaled = self.min_val + (self.max_val - self.min_val) * mu
        phi = inputs[:, 1:2]  # Keep phi unchanged
        return tf.concat([mu_scaled, phi], axis=1)
    
    def get_config(self):
        config = super().get_config()
        config.update({
            "min_val": self.min_val,
            "max_val": self.max_val
        })
        return config

def mae_mu(y_true, y_pred):
    mu = y_pred[:, 0]  
    return tf.reduce_mean(tf.abs(y_true - mu))

HP = {
    "lr": 1e-3,                
    "weight_decay": 1e-5, 
    "l2_reg": 1e-5,            
    "units1": 128,             
    "units2": 64,            
    "dropout": 0.10,          
    "batch_size": 16,          
    "patience_es": 12,        
    "patience_rlr": 6,         
}

def create_beta_model(input_dim, model_name, hp=HP, target_min=0.0, target_max=1.0):
    inputs = layers.Input(shape=(input_dim,))
    x = layers.Dense(hp["units1"], activation=tf.keras.layers.LeakyReLU(alpha=0.1),
                    kernel_regularizer=regularizers.l2(hp["l2_reg"]))(inputs)
    x = LayerNormalization()(x)
    x = layers.Dropout(0.10)(x)

    x = layers.Dense(hp["units2"], activation=tf.keras.layers.LeakyReLU(alpha=0.1),
                    kernel_regularizer=regularizers.l2(hp["l2_reg"]))(x)
    x = LayerNormalization()(x)
    x = layers.Dropout(0.10)(x)
    
    # Raw outputs (0-1 range)
    mu_raw = layers.Dense(1, activation='sigmoid', name=f'{model_name}_mu_raw')(x)
    phi = layers.Dense(1, activation='softplus', name=f'{model_name}_phi')(x)
    
    # Combine raw outputs
    raw_outputs = layers.Concatenate(name=f'{model_name}_raw_output')([mu_raw, phi])
    
    outputs = OutputScaler(min_val=target_min, max_val=target_max, 
                          name=f'{model_name}_output')(raw_outputs)

    model = keras.Model(inputs, outputs)
    optimizer = tf.keras.optimizers.AdamW(learning_rate=HP["lr"] , weight_decay=HP["weight_decay"])
    
    model.compile(
        optimizer=optimizer,
        loss=beta_nll_loss,
        metrics=[mae_mu]
    )
    
    return model 

input_dim = x_train_scaled.shape[1]

# Calculate actual ranges from your training data
fert_min, fert_max = y_fert_train.min(), y_fert_train.max()
blast_min, blast_max = y_blast_train.min(), y_blast_train.max()

print(f"Fertilization target range: {fert_min:.3f} - {fert_max:.3f}")
print(f"Blastulation target range: {blast_min:.3f} - {blast_max:.3f}")

# Create models with appropriate scaling
fert_model = create_beta_model(input_dim, "fert", target_min=fert_min, target_max=fert_max)
blast_model = create_beta_model(input_dim, "blast", target_min=blast_min, target_max=blast_max)

print("Fertilization Model Summary:")
fert_model.summary()
print("\nBlastulation Model Summary:")
blast_model.summary()
```

```python
##Implementing K-fold cross validation 
k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)

fert_val_mae_scores = []

print(f"\n--- Performing {k}-Fold Cross Validation (Fertilisation Model) ---\n")

fold = 1
for train_idx, val_idx in kf.split(x_train_scaled):
    print(f"Fold {fold}/{k}")

    # Split data for this fold
    X_train_fold, X_val_fold = x_train_scaled[train_idx], x_train_scaled[val_idx]
    y_fert_train_fold, y_fert_val_fold = y_fert_train[train_idx], y_fert_train[val_idx]

    # Create new model for each fold
    fert_model = create_beta_model(input_dim, "fert", target_min=fert_min, target_max=fert_max)

    fert_callbacks = [
        EarlyStopping(monitor='val_loss', patience=HP["patience_es"], restore_best_weights=True),
        ReduceLROnPlateau(monitor='val_loss', patience=HP["patience_rlr"], factor=0.5, min_lr=1e-6)
    ]

    fert_model.fit(
        X_train_fold, y_fert_train_fold,
        validation_data=(X_val_fold, y_fert_val_fold),
        epochs=100,
        batch_size=HP["batch_size"],
        callbacks=fert_callbacks,
        verbose=0
    )

    fert_eval = fert_model.evaluate(X_val_fold, y_fert_val_fold, verbose=0)
    fert_val_mae_scores.append(fert_eval[1])  # mae_mu index

    print(f"  Fold {fold} Fert MAE_mu: {fert_eval[1]:.4f}")
    fold += 1

# --- Summarize Cross-Validation Results ---
# Cross-validation performed only on fertilisation model
# to assess base predictive stability before conditional modelling
print("\nCross-Validation Results:")
print(f"Fertilisation MAE_mu (mean ± std): {np.mean(fert_val_mae_scores):.4f} ± {np.std(fert_val_mae_scores):.4f}")
print("------------------------------------------------------")


# --- Retrain final fertilisation model on full training data ---
print("\nRetraining fertilisation model on full training data...")

fert_model = create_beta_model(input_dim, "fert", target_min=fert_min, target_max=fert_max)
fert_callbacks = [
    EarlyStopping(monitor='val_loss', patience=HP["patience_es"], restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', patience=HP["patience_rlr"], factor=0.5, min_lr=1e-6)
]

fert_model.fit(
    x_train_scaled, y_fert_train,
    validation_data=(x_val_scaled, y_fert_val),
    epochs=100,
    batch_size=HP["batch_size"],
    callbacks=fert_callbacks,
    verbose=1
)

# --- Combine training and validation sets once ---
x_full_scaled = np.concatenate([x_train_scaled, x_val_scaled], axis=0)
y_blast_full = np.concatenate([y_blast_train, y_blast_val], axis=0)

# --- Predict fertilisation rate for exactly the same rows ---
fert_pred_full = fert_model.predict(x_full_scaled)[:, 0]
print(f"x_full_scaled: {x_full_scaled.shape}, fert_pred_full: {fert_pred_full.shape}")

# --- Filter for non-zero fertilisation ---
valid_idx = fert_pred_full > 0

# --- Apply same mask across all arrays ---
x_blast = x_full_scaled[valid_idx]
y_blast = y_blast_full[valid_idx]
fert_pred_valid = fert_pred_full[valid_idx]

# --- Prepare data for blastulation model --

# Enrich with predicted fertilisation as an input
x_blast_enriched = np.concatenate([x_blast, fert_pred_valid.reshape(-1, 1)], axis=1)

print(f"\nTraining Blastulation Model on {x_blast_enriched.shape[0]} samples (fert > 0)")

# --- Train Blastulation Model ---
blast_model = create_beta_model(
    input_dim=x_blast_enriched.shape[1],
    model_name="blast",
    target_min=blast_min,
    target_max=blast_max
)

blast_callbacks = [
    EarlyStopping(monitor='val_loss', patience=HP["patience_es"], restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', patience=HP["patience_rlr"], factor=0.5, min_lr=1e-6)
]

blast_model.fit(
    x_blast_enriched, y_blast,
    validation_split=0.2,
    epochs=100,
    batch_size=HP["batch_size"],
    callbacks=blast_callbacks,
    verbose=1
)

print("\n✅ Both models trained successfully!")
```

```python
def create_beta_model_shallow(input_dim, model_name, target_min, target_max, hp=HP):
    
    inputs = layers.Input(shape=(input_dim,))
    
    # Only ONE hidden layer
    x = layers.Dense(
        hp["units2"],
        activation=tf.keras.layers.LeakyReLU(negative_slope=0.1),
        kernel_regularizer=regularizers.l2(hp["l2_reg"])
    )(inputs)
    
    x = LayerNormalization()(x)
    x = layers.Dropout(hp["dropout"])(x)
    
    # Output layers
    mu_raw = layers.Dense(1, activation='sigmoid', name=f'{model_name}_mu_raw')(x)
    phi = layers.Dense(1, activation='softplus', name=f'{model_name}_phi')(x)
    
    raw_outputs = layers.Concatenate()([mu_raw, phi])
    
    outputs = OutputScaler(
        min_val=target_min,
        max_val=target_max
    )(raw_outputs)
    
    model = keras.Model(inputs, outputs)
    
    optimizer = tf.keras.optimizers.AdamW(
        learning_rate=hp["lr"],
        weight_decay=hp["weight_decay"]
    )
    
    model.compile(
        optimizer=optimizer,
        loss=beta_nll_loss,
        metrics=[mae_mu]
    )
    
    return model
```

```python
def compare_full_vs_shallow(
    x_train, y_train,
    x_val, y_val,
    input_dim,
    target_min,
    target_max,
    n_runs=10
):
    
    full_mae = []
    shallow_mae = []
    
    print("\nRunning FULL model experiments...")
    
    for i in range(n_runs):
        
        model = create_beta_model(
            input_dim=input_dim,
            model_name="fert_full",
            target_min=target_min,
            target_max=target_max
        )
        
        model.fit(
            x_train, y_train,
            validation_data=(x_val, y_val),
            epochs=100,
            batch_size=HP["batch_size"],
            verbose=0,
            callbacks=[
                EarlyStopping(
                    patience=HP["patience_es"],
                    restore_best_weights=True
                )
            ]
        )
        
        pred = model.predict(x_val)[:, 0]
        mae = mean_absolute_error(y_val, pred)
        full_mae.append(mae)
        
        print(f"Full Run {i+1}: MAE = {mae:.4f}")
    
    
    print("\nRunning SHALLOW model experiments...")
    
    for i in range(n_runs):
        
        model = create_beta_model_shallow(
            input_dim=input_dim,
            model_name="fert_shallow",
            target_min=target_min,
            target_max=target_max
        )
        
        model.fit(
            x_train, y_train,
            validation_data=(x_val, y_val),
            epochs=100,
            batch_size=HP["batch_size"],
            verbose=0,
            callbacks=[
                EarlyStopping(
                    patience=HP["patience_es"],
                    restore_best_weights=True
                )
            ]
        )
        
        pred = model.predict(x_val)[:, 0]
        mae = mean_absolute_error(y_val, pred)
        shallow_mae.append(mae)
        
        print(f"Shallow Run {i+1}: MAE = {mae:.4f}")
    
    
    print("\n========== FINAL COMPARISON ==========")
    print(f"Full Model:    mean={np.mean(full_mae):.4f}, std={np.std(full_mae):.4f}")
    print(f"Shallow Model: mean={np.mean(shallow_mae):.4f}, std={np.std(shallow_mae):.4f}")
    
    return full_mae, shallow_mae
```

```python
# ===============================
# 🔍 SENSITIVITY ANALYSIS (ADD HERE)
# ===============================

def sensitivity_analysis(model, x , feature_names, delta=0.5):
    baseline_pred = model.predict(x)[:,0]
    sensitivities=[]

    for i, fname in enumerate(feature_names):
        x_plus = x.copy()
        x_minus=x.copy()

        x_plus[:,i] += delta
        x_minus[:,i] -= delta

        pred_plus = model.predict(x_plus)[:,0]
        pred_minus = model.predict(x_minus)[:,0]

        sens = np.mean(np.abs(pred_plus - pred_minus))/ (2*delta)
        sensitivities.append(sens)

        
    return pd.DataFrame({
            "Feature": feature_names, 
            "Sensitivity": sensitivities
        }).sort_values("Sensitivity", ascending=False)
    
feature_names = x_train.columns.tolist()

fert_sensitivity_df = sensitivity_analysis(
    model = fert_model, 
    x= x_val_scaled, 
    feature_names= feature_names, 
    delta=0.5
)

print("\n Top Fert Sensitivities")
print(fert_sensitivity_df.head(10))

plt.figure(figsize=(8,6))
plt.barh(
    fert_sensitivity_df["Feature"],
    fert_sensitivity_df["Sensitivity"]
)
plt.gca().invert_yaxis()
plt.xlabel("Mean |Δ Prediction| per 1 SD change")
plt.title("Sensitivity Analysis – Fertilization Model")
plt.tight_layout()
plt.show()
```

```python
def average_sensitivity_over_runs(
    create_model_fn,
    x_train,
    y_train,
    x_val,
    y_val,
    feature_names,
    target_min,
    target_max,
    n_runs=10,
    delta=0.5
):
    all_sensitivities = []

    for run in range(n_runs):

        print(f"\nSensitivity Run {run+1}/{n_runs}")

        model = create_model_fn(
            input_dim=x_train.shape[1],
            model_name=f"fert_run_{run}",
            target_min=target_min,
            target_max=target_max
        )

        model.fit(
            x_train, y_train,
            validation_data=(x_val, y_val),
            epochs=100,
            batch_size=HP["batch_size"],
            verbose=0,
            callbacks=[
                EarlyStopping(
                    patience=HP["patience_es"],
                    restore_best_weights=True
                )
            ]
        )

        sens_df = sensitivity_analysis(
            model=model,
            x=x_val,
            feature_names=feature_names,  # ✅ FIXED
            delta=delta
        )

        all_sensitivities.append(sens_df["Sensitivity"].values)

    all_sensitivities = np.array(all_sensitivities)

    mean_sens = np.mean(all_sensitivities, axis=0)
    std_sens = np.std(all_sensitivities, axis=0)

    result_df = pd.DataFrame({
        "Feature": feature_names,  # ✅ FIXED
        "Mean Sensitivity": mean_sens,
        "Std Sensitivity": std_sens
    })

    result_df = result_df.sort_values(
        "Mean Sensitivity",
        ascending=False
    )

    return result_df

fert_sensitivity_avg = average_sensitivity_over_runs(
    create_model_fn=create_beta_model,
    x_train=x_train_scaled,
    y_train=y_fert_train,
    x_val=x_val_scaled,
    y_val=y_fert_val,
    feature_names=x_train.columns.tolist(),
    target_min=fert_min,
    target_max=fert_max,
    n_runs=10
)

print("\nFINAL averaged fertilization sensitivity:")
print(fert_sensitivity_avg)
fert_sensitivity_avg.to_csv("fertilization_sensitivity_results.csv")
```

```python

def blast_sensitivity_analysis(
    blast_model,
    fert_model,
    X,
    feature_names,
    delta=0.5
):
    fert_pred = fert_model.predict(X)[:, 0]
    X_base = np.concatenate([X, fert_pred.reshape(-1,1)], axis=1)

    baseline_pred = blast_model.predict(X_base)[:, 0]
    sensitivities = []

    for i, fname in enumerate(feature_names):
        X_plus = X.copy()
        X_minus = X.copy()

        X_plus[:, i] += delta
        X_minus[:, i] -= delta

        fert_plus = fert_model.predict(X_plus)[:, 0]
        fert_minus = fert_model.predict(X_minus)[:, 0]

        X_plus_enriched = np.concatenate([X_plus, fert_plus.reshape(-1,1)], axis=1)
        X_minus_enriched = np.concatenate([X_minus, fert_minus.reshape(-1,1)], axis=1)

        pred_plus = blast_model.predict(X_plus_enriched)[:, 0]
        pred_minus = blast_model.predict(X_minus_enriched)[:, 0]

        sens = np.mean(np.abs(pred_plus - pred_minus)) / (2 * delta)
        sensitivities.append(sens)

    return pd.DataFrame({
        "Feature": feature_names,
        "Sensitivity": sensitivities
    }).sort_values("Sensitivity", ascending=False)


blast_sensitivity_df = blast_sensitivity_analysis(
    blast_model=blast_model,
    fert_model=fert_model,
    X=x_val_scaled,
    feature_names=feature_names,
    delta=0.5
)

print("\nTop Blastulation Sensitivities:")
print(blast_sensitivity_df.head(10))

plt.figure(figsize=(8,6))
plt.barh(
    blast_sensitivity_df["Feature"],
    blast_sensitivity_df["Sensitivity"]
)
plt.gca().invert_yaxis()
plt.xlabel("Mean |Δ Prediction| per 1 SD change")
plt.title("Sensitivity Analysis – Blastulation Model")
plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, regularizers, Model


def create_resnet_model(input_dim , model_name, hp=HP, target_min=0.0, target_max=1.0 , n_blocks=3): 
    inputs = layers.Input(shape=(input_dim,))
    # First dense layer
    x = layers.Dense(hp["units1"], activation='relu', kernel_regularizer=regularizers.l2(hp["l2_reg"]))(inputs)

    # Residual blocks
    for _ in range(n_blocks):
        shortcut = x  # Save input for residual connection
        x = layers.Dense(hp["units1"], activation='relu', kernel_regularizer=regularizers.l2(hp["l2_reg"]))(x)
        x = layers.Dense(hp["units1"], activation=None, kernel_regularizer=regularizers.l2(hp["l2_reg"]))(x)
        x = layers.Add()([shortcut, x])  # Residual connection
        x = layers.Activation('relu')(x)
        x = layers.BatchNormalization()(x)

    # Output layer
    mu_raw = layers.Dense(1, activation='sigmoid', name=f'{model_name}_mu_raw')(x)
    phi = layers.Dense(1, activation='softplus', name=f'{model_name}_phi')(x)
    
    raw_outputs = layers.Concatenate()([mu_raw, phi])
    
    outputs = OutputScaler(min_val=target_min, max_val=target_max)(raw_outputs)

    model = Model(inputs, outputs)

    optimizer = tf.keras.optimizers.AdamW(
        learning_rate=hp["lr"],
        weight_decay=hp["weight_decay"]
    )

    model.compile(
        optimizer=optimizer,
        loss=beta_nll_loss,
        metrics=[mae_mu]
    )
    
    return model


resnet_model = create_resnet_model(
    input_dim=input_dim,       # same as your other models
    model_name="fert_resnet",  # a descriptive name
    hp=HP,                     # your hyperparameter dict
    target_min=fert_min,       # same as before
    target_max=fert_max,       # same as before
    n_blocks=3                 # number of residual blocks
)

class PrintMAE(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        print(f"Epoch {epoch+1}: MAE_mu = {logs['mae_mu']:.4f}, val_MAE_mu = {logs['val_mae_mu']:.4f}")


    
```

```python
# 10-run evaluation loop
n_runs = 10
mae_list = []

print("Running 10 ResNet experiments...")

for i in range(n_runs):
    print(f"\nRun {i+1}/{n_runs}")
    model = create_resnet_model(input_dim=input_dim, model_name="fert_resnet", hp=HP,
                                target_min=fert_min, target_max=fert_max, n_blocks=3)
    
    model.fit(
        x_train_scaled, y_fert_train,
        validation_data=(x_val_scaled, y_fert_val),
        epochs=100,
        batch_size=HP["batch_size"],
        verbose=0,
        callbacks=[EarlyStopping(patience=HP["patience_es"], restore_best_weights=True),
                   PrintMAE()]
    )
    
    pred_val = model.predict(x_val_scaled)[:,0]
    mae_val = mean_absolute_error(y_fert_val, pred_val)
    mae_list.append(mae_val)
    
    print(f"Run {i+1} validation MAE: {mae_val:.4f}")

# Final statistics
mean_mae = np.mean(mae_list)
std_mae = np.std(mae_list)
print("\n========== FINAL RESNET COMPARISON ==========")
print(f"Mean MAE: {mean_mae:.4f}, Std MAE: {std_mae:.4f}")
```
