```python
# -------------------------------------
# Load data
# -------------------------------------

df = pd.read_csv(
    'RBL Data Edited - Combined (2019, 20, 21, 23. 24).csv',
    sep=";",
    decimal=",",
    encoding="utf-8",
    usecols=lambda col: col != 'Patient_ID'
)

# Remove completely empty rows
df = df.dropna(how="all")

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# -------------------------------------
# Missingness report
# -------------------------------------

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
}).sort_values("Missing_Percentage", ascending=False)

print(missing_summary)

# -------------------------------------
# Remove rows with missing targets
# -------------------------------------

df = df.dropna(
    subset=[
        "Fertilization Rate (%)",
        "Blastulation_Rate (%)"
    ]
)

# -------------------------------------
# Convert targets from %
# -------------------------------------

for col in [
    "Fertilization Rate (%)",
    "Blastulation_Rate (%)"
]:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("%", "", regex=False)
    )

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    ) / 100

# -------------------------------------
# Features and targets
# -------------------------------------

x = df.drop(
    columns=[
        "Fertilization Rate (%)",
        "Blastulation_Rate (%)",
        "Year",
        "Ejaculatory_Abstinence (days)"
    ]
)

y = df[
    [
        "Fertilization Rate (%)",
        "Blastulation_Rate (%)"
    ]
]

# -------------------------------------
# Numeric and categorical columns
# -------------------------------------

num_cols = x.select_dtypes(
    include=["float64", "int64"]
).columns.tolist()

cat_cols = x.select_dtypes(
    include=["object"]
).columns.tolist()

# -------------------------------------
# Standardize categorical variables
# -------------------------------------

cat_cols = x.select_dtypes(include=["object"]).columns

for col in cat_cols:
    x[col] = (
        x[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Fix known spelling issues
x["Male_Infertility"] = x["Male_Infertility"].replace({
    "secondayr": "secondary"
})

# Convert unknown values to NaN
missing_values = [
    "unknown",
    "",
    " "
]

for col in cat_cols:
    x[col] = x[col].replace(missing_values, np.nan)
    
# -------------------------------------
# Imputation
# -------------------------------------

num_imputer = IterativeImputer(
    random_state=42,
    max_iter=20
)

cat_imputer = SimpleImputer(
    strategy="most_frequent"
)

x[num_cols] = num_imputer.fit_transform(
    x[num_cols]
)

x[cat_cols] = cat_imputer.fit_transform(
    x[cat_cols]
)

# -------------------------------------
# Label Encoding
# -------------------------------------
encoded_df = x.copy()
le_col = {}

for col in cat_cols:
    le = LabelEncoder()

    x[col] = le.fit_transform(
        x[col].astype(str)
    )

    le_col[col] = le

encoded_df = x.copy()

# -------------------------------------
# Validation Checks
# -------------------------------------

print("\nFeature Null Counts:")
print(x.isnull().sum())

print("\nTarget Null Counts:")
print(y.isnull().sum())

print("\nFeature Shape:")
print(x.shape)

print("\nTarget Shape:")
print(y.shape)

print("\nTotal Features:")
print(len(x.columns))
```

```python
# Safe conversion for Fertilization
df["Fertilization Rate (%)"] = pd.to_numeric(
    df["Fertilization Rate (%)"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

# Safe conversion for Blastulation
df["Blastulation_Rate (%)"] = pd.to_numeric(
    df["Blastulation_Rate (%)"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

df["Blastulation_Rate (%)"] = pd.to_numeric(
    df["Blastulation_Rate (%)"],
    errors="coerce"
)

print(df.shape)
print(df[["Male_Age (years)", "Female_Age (years)"]].isnull().sum())

print(df[["Fertilization Rate (%)","Blastulation_Rate (%)"]].dtypes)
print(df[["Fertilization Rate (%)","Blastulation_Rate (%)"]].isnull().sum())
```

```python
# -------------------------------------
# Descriptive statistics for reporting
# -------------------------------------
from scipy.stats import shapiro

# Create reporting copy with targets in percentage scale
df_report = df.copy()
df_report["Fertilization Rate (%)"] = df_report["Fertilization Rate (%)"] * 100
df_report["Blastulation_Rate (%)"] = df_report["Blastulation_Rate (%)"] * 100

# Add Follicles Collected from x (not in df after feature split)
df_report["Follicles_Collected (#)"] = x["Follicles_Collected (#)"].values

# Add Round Cells from original df for honest missingness reporting
# (only 4 non-missing values — do not use imputed version)
table_vars = [
    "Fertilization Rate (%)",
    "Blastulation_Rate (%)",
    "Male_Age (years)",
    "Female_Age (years)",
    "Semen_Volume (ml)",
    "Sperm_Concentration (million/ml)",
    "Sperm_Prog_Motility (%)",
    "Sperm_Morphology (%)",
    "Round_Cells (million/ml)",
    "Follicles_Collected (#)"
]

summary = []
for col in table_vars:
    vals = pd.to_numeric(df_report[col], errors="coerce").dropna()
    n = len(vals)
    mean = vals.mean()
    sd = vals.std()
    median = vals.median()
    q1 = vals.quantile(0.25)
    q3 = vals.quantile(0.75)
    min_val = vals.min()
    max_val = vals.max()

    if n >= 8:
        sample = vals if n <= 5000 else vals.sample(5000, random_state=42)
        stat, p = shapiro(sample)
        distribution = "Parametric" if p > 0.05 else "Non-parametric"
    else:
        stat, p = None, None
        distribution = "Insufficient sample"

    summary.append({
        "Variable": col,
        "n": n,
        "Mean ± SD": f"{mean:.1f} ± {sd:.1f}",
        "Median (IQR)": f"{median:.1f} ({q1:.1f}–{q3:.1f})",
        "Range": f"{min_val:.1f}–{max_val:.1f}",
        "Distribution": distribution
    })

table4 = pd.DataFrame(summary)
print(table4.to_string())
table4.to_csv("summary_statistics_table4.csv", index=False)

# -------------------------------------
# Missingness report for text reporting
# -------------------------------------
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df) * 100).round(1)
}).sort_values("Missing_Percentage", ascending=False)
missing_report = missing_report[missing_report["Missing_Count"] > 0]
print("\nMissing data summary:")
print(missing_report.to_string())
```
